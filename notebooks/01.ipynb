```python
import pandas as pd
```

```python
files = {
    "sales_validation": "sales_train_validation.csv",
    "sales_evaluation": "sales_train_evaluation.csv",
    "calendar": "calendar.csv",
    "prices": "sell_prices.csv",
}

dfs = {}

for name, path in files.items():
    df = pd.read_csv(path)
    dfs[name] = df
    
    print(f"\n{name}")
    print("shape:", df.shape)
    print("memory MB:", round(df.memory_usage(deep=True).sum() / 1024**2, 2))
    display(df.head())
```

```python
for name, df in dfs.items():
    print(f"\n--- {name} ---")
    print(df.dtypes)
```

```python
for name, df in dfs.items():
    print(f"\n--- {name} missing values ---")
    
    missing = (
        df.isna()
        .sum()
        .sort_values(ascending=False)
    )
    
    display(missing[missing > 0].head(20))
```

```python
sales = dfs["sales_evaluation"]

print("stores:", sales["store_id"].nunique())
print("items:", sales["item_id"].nunique())
print("departments:", sales["dept_id"].nunique())
print("categories:", sales["cat_id"].nunique())
print("states:", sales["state_id"].unique())

print("\nStores:")
print(sales["store_id"].value_counts())
```

```python
day_cols = [c for c in sales.columns if c.startswith("d_")]

print("number of sales days:", len(day_cols))
print("first:", day_cols[:5])
print("last:", day_cols[-5:])
```
